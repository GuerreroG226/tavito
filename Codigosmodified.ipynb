# **PRIMER VIDEO:**

**LISTAS**

```python
lista_frutas = ["manzana", "banana", "cereza", "dátil", "fresa"]

print(lista_frutas)
print(lista_frutas[-1]) # último elemento
print(lista_frutas[1:4]) # porción del índice 1 al 3

lista_frutas[2] = "uva" # cambiar un elemento
print(lista_frutas)

lista_frutas.append("kiwi") # añadir un elemento
print(lista_frutas)

print(len(lista_frutas))
```

**TUPLAS**

```python
tupla_colores = ("rojo", "verde", "azul", "rojo")
print(tupla_colores)

print(tupla_colores[0])
print(tupla_colores[1])
print(tupla_colores[2])

print(tupla_colores.count("rojo")) # contar ocurrencias
print(tupla_colores.index("azul")) # índice de un elemento

print(("amarillo")) # ejemplo de tupla de un solo elemento (no)
print(("amarillo",)) # tupla de un solo elemento correctamente
```

**CONJUNTOS**

```python
print(set())

print(set(["manzana", "banana", "manzana", "uva", "kiwi"]))
print(set(("pera", "mango", "pera", "limón", "mango")))
print(set("sandia")) # conjunto a partir de una cadena

conjunto_frutas_1 = set(["manzana", "banana", "cereza"])
conjunto_frutas_2 = set(["banana", "dátil", "fresa"])
conjunto_frutas_3 = set(["manzana", "cereza"])

print(conjunto_frutas_1)
print(conjunto_frutas_2)
print(conjunto_frutas_3)

conjunto_frutas_1.add("kiwi")
print(conjunto_frutas_1)

conjunto_frutas_1.remove("banana")
print(conjunto_frutas_1)

print(conjunto_frutas_1.intersection(conjunto_frutas_2))
print(conjunto_frutas_3.issubset(conjunto_frutas_1))
print(conjunto_frutas_2.issubset(conjunto_frutas_1))
```

**DICCIONARIO**

```python
diccionario_colores = {"manzana": "rojo", "banana": "amarillo"}
diccionario_colores["uva"] = "morado"
print(diccionario_colores)

dict_frutas_precios = dict([("manzana", 1.50), ("banana", 0.75), ("uva", 2.20)])
print(dict_frutas_precios)

dict_frutas_cantidades = dict(manzana = 10, banana = 15, uva = 5)
print(dict_frutas_cantidades)

dict_info_frutas = {"manzana": {"color": "rojo", "sabor": "dulce"},
                   "banana": {"color": "amarillo", "sabor": "dulce"}}

print(dict_info_frutas)

dict_fruta_favorita = {"ana": "manzana", "ana": "uva"} # La clave "ana" se sobrescribirá
print(dict_fruta_favorita)

print(diccionario_colores,
      diccionario_colores.keys(),
      diccionario_colores.values(),
      diccionario_colores.items())

valores_colores = diccionario_colores.values()
print(valores_colores)
diccionario_colores["manzana"] = "verde"
print(valores_colores)
diccionario_colores.pop("banana")
print(diccionario_colores)
```

# **SEGUNDO VIDEO:**

```python
# Ejemplo de conversión de distintos tipos a False

print(0, "->", bool(0))
print(0.0, "->", bool(0.0))
print("", "->", bool(""))
print([], "->", bool([]))
print((), "->", bool(()))
print(set(), "->", bool(set()))
print({}, "->", bool({}))
print(None, "->", bool(None))
print(False, "->", bool(False))
```

```python
# Ejemplo de conversión de distintos tipos a True

print(100, "->", bool(100))
print(-5.0, "->", bool(-5.0))
print("Hola Mundo", "->", bool("Hola Mundo"))
print([0], "->", bool([0]))
print(("a",), "->", bool(("a",)))
print({1}, "->", bool({1}))
print({1: "one"}, "->", bool({1: "one"}))
```

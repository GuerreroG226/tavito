# **ESTRUCTURA DE DATOS EN PYTHON**

LISTAS

```python
lista = [29, True, 3.1415, "El número de Avogadro sí que mola"]

print(lista)
print(lista[-1])
print(lista[1:3])

lista[2] = "He cambiado este elemento"

lista[2] = [3, 2, 1]
print(lista)

print(len(lista))
```

```python
lista_nueva = [1, 2, 3, 4, 5]

lista_nueva.append(3)

print(lista_nueva)

print(lista_nueva.count(3))

print(lista_nueva.index(4))

lista_nueva.remove(3)

print(lista_nueva)
```

TUPLAS

```python
tupla = ("¿La tierra es plana?", True, False)
print(tupla)

print(tupla[0])
print(tupla[1])
print(tupla[2])

print(tupla.count(True))
print(tupla.index(False))

print((1))
print((1,))
```

CONJUNTOS

```python
print(set())

print(set([5, 2, 5, 1, 1.5]))
print(set((5, 2, 5, 1, 1.5)))
print(set("52511.5"))

conjunto = set([2, 3, 3, 4])
conjunto_2 = set([5, 3, 5, 6])
conjunto_3 = set([4, 2])

print(conjunto)
print(conjunto_2)
print(conjunto_3)

conjunto.add(1)
print(conjunto)

conjunto.remove(1)
print(conjunto)

print(conjunto.intersection(conjunto_2))
print(conjunto_2.issubset(conjunto))
print(conjunto_3.issubset(conjunto))
```

DICCIONARIOS

```python
diccionario = {1: "Uno", 2: "Dos"}
diccionario[3] = "Tres"
print(diccionario)

dict_lista_tuplas = dict([(1, "Uno"), (2, "Dos"), (3, "Tres")])
print(dict_lista_tuplas)

dict_lista_string = dict(Uno = 1, Dos = 2, Tres = 3)
print(dict_lista_string)

dict_tipos = {1: "integer",
              2.2: "float",
              "texto": "string",
              (1, 2): "tupla"}

print(dict_tipos)

dict_repeticion = {1: "Primero", 1: "Último"}
print(dict_repeticion)

print(diccionario,
      diccionario.keys(),
      diccionario.values(),
      diccionario.items())

claves = diccionario.values()
print(claves)
diccionario[1] = "One"
print(claves)
diccionario.pop(2)
print(diccionario)
```

```python
# Ejemplo de conversión de distintos tipos a False

print(0, "->", bool(0))
print(0.0, "->", bool(0.0))
print("", "->", bool(""))
print(list(), "->", bool(list()))
print(tuple(), "->", bool(tuple()))
print(set(), "->", bool(set()))
print(dict(), "->", bool(dict()))
print(None, "->", bool(None))
```

```python
# Ejemplo de conversión de distintos tipos a True

print(9000, "->", bool(9000))
print(3.1415, "->", bool(3.1415))
print("Cadena", "->", bool("Cadena"))
print(list((1, 2)), "->", bool(list((1, 2))))
print(tuple((1, 2)), "->", bool(tuple((1, 2))))
print(set((1, 2)), "->", bool(set((1, 2))))
print(dict([(1,2)]), "->", bool(dict([(1,1)])))
```

```python
# Ejemplo if-else

nota = float(input("Introduzca su nota: "))
if nota < 5:
  print("Estás suspenso, F")
else:
  print("Estás aprobado, ¡enhorabuena!")
```

```python
#Ejemplo if-elif-else

nota = float(input("Introduzca su nota: "))
if 0 <= nota and nota < 5:
  print("Estás suspenso, F")
elif 5 <= nota and nota < 7:
  print("Aprobado")
elif 7 <= nota and nota < 9:
  print("Notable")
elif 9 <= nota and nota <= 10:
  print("Sobresaliente")
else:
  print("Nota fuera de rango")
```

```python
# Ejemplo while

contador = 3
while contador != 0:
  contador = contador - 1
  print("El contenido de contador es", contador)
```

```python
# Acceso por índices a una lista con un for

lista = ["Vamos", "a", "acceder", "a", "esta", "lista", "por", "índices"]

for indice in range(len(lista)):
  print("Indice:", indice,"Elemento:", lista[indice])
```

```python
# for para visualizar el efecto de las sentencias break y continue

for numero in range(5):
  print(numero)
```

```python
# Visualización del efecto de la sentencia break

for numero in range(5):
  if numero == 3:
    break
  print(numero)
```

```python
# Visualización del efecto de la sentencia continue

for numero in range(5):
  if numero == 3:
    continue
  print(numero)
```

```python
# Uso del else en un while

intentos = 3
while intentos > 0:
  if input(">>> Escriba la contraseña:") == "BitBoss":
    print("¡Correcta!")
    break
  intentos = intentos - 1
  print("Contraseña incorrecta")
else:
  print("Se te acabaron los intentos :(")
```
